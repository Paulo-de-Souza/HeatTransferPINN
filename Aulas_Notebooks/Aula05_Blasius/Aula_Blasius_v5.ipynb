# **Solução de Blasius: camada-limite hidrodinâmica e térmica**

Neste notebook estudaremos a solução de Blasius para o escoamento laminar sobre uma placa plana.

A análise será dividida em duas partes:

1. solução da camada-limite hidrodinâmica;
2. solução da camada-limite térmica.

Primeiramente, resolveremos a equação de Blasius por métodos numéricos clássicos. Em seguida, estudaremos como o mesmo campo de velocidade influencia a distribuição de temperatura próxima à parede.

## **Por que estudar a camada-limite?**

Quando um fluido escoa sobre uma superfície sólida, a condição de não deslizamento impõe que a velocidade do fluido seja igual à velocidade da parede. Para uma placa fixa, isso significa que $u=0$ junto à superfície.

Longe da placa, o escoamento praticamente não sente a presença da parede e a velocidade se aproxima da velocidade externa $U_\infty$.

Entre essas duas regiões existe uma faixa relativamente fina na qual a velocidade varia rapidamente. Essa região é chamada de **camada-limite hidrodinâmica**.

Conhecer essa região é importante porque nela aparecem os maiores gradientes de velocidade. Esses gradientes estão relacionados ao cisalhamento na parede e, consequentemente, à força de arrasto sobre a superfície.

Além disso, o escoamento próximo à parede transporta energia térmica. Portanto, a estrutura da camada hidrodinâmica também influencia a transferência de calor entre a superfície e o fluido.

## **Camada hidrodinâmica e camada térmica**

A **camada-limite hidrodinâmica** descreve como a velocidade varia desde o valor imposto pela parede até a velocidade do escoamento externo.

Essa informação permite estudar, por exemplo:

- o gradiente de velocidade na parede;
- a tensão de cisalhamento;
- o arrasto sobre a placa;
- o desenvolvimento do escoamento ao longo da superfície.

A **camada-limite térmica** descreve como a temperatura varia entre a temperatura da parede e a temperatura do escoamento externo.

Ela permite estudar:

- o gradiente de temperatura na parede;
- o fluxo de calor;
- o coeficiente convectivo;
- o número de Nusselt.

As duas camadas não precisam ter a mesma espessura. Essa diferença depende principalmente do número de Prandtl, que compara a difusão de quantidade de movimento com a difusão térmica.

## **Hipóteses adotadas**

Para obter uma solução simples e didática, consideraremos:

- escoamento permanente;
- escoamento bidimensional;
- fluido incompressível;
- propriedades constantes;
- fluido newtoniano;
- escoamento laminar;
- placa plana;
- gradiente de pressão externo nulo;
- velocidade externa uniforme $U_\infty$;
- ausência de separação da camada-limite.

Essas hipóteses permitem transformar o problema original, que depende das coordenadas $x$ e $y$, em um problema baseado em uma única variável de similaridade.

## **Da equação de camada-limite à variável de similaridade**

Para um escoamento permanente, incompressível e bidimensional sobre uma placa plana, as equações de conservação de massa e quantidade de movimento podem ser escritas, sob as hipóteses da teoria de camada-limite, como

$$
\frac{\partial u}{\partial x}
+
\frac{\partial v}{\partial y}
=0,
$$

e

$$
u\frac{\partial u}{\partial x}
+
v\frac{\partial u}{\partial y}
=
\nu\frac{\partial^2 u}{\partial y^2}.
$$

A primeira equação representa a conservação de massa. A segunda descreve o balanço entre a advecção da quantidade de movimento e sua difusão viscosa.

As condições de contorno são

$$
u(x,0)=0,
\qquad
v(x,0)=0,
$$

na parede, devido à condição de não deslizamento e à impermeabilidade da placa, e

$$
u(x,y)\longrightarrow U_\infty
\qquad
\text{quando } y\longrightarrow\infty.
$$

Ou seja, a velocidade deve ser nula junto à placa e se aproximar da velocidade do escoamento externo longe dela.

### **Por que aparece a variável $\eta$?**

A camada-limite possui uma espessura característica $\delta(x)$, que cresce ao longo da placa.

Podemos estimar essa espessura comparando os termos dominantes da equação de quantidade de movimento:

$$
u\frac{\partial u}{\partial x}
\sim
\frac{U_\infty^2}{x},
$$

enquanto o termo de difusão viscosa possui ordem de grandeza

$$
\nu\frac{\partial^2u}{\partial y^2}
\sim
\nu\frac{U_\infty}{\delta^2}.
$$

Igualando essas duas escalas:

$$
\frac{U_\infty^2}{x}
\sim
\nu\frac{U_\infty}{\delta^2},
$$

obtemos

$$
\delta(x)
\sim
\sqrt{\frac{\nu x}{U_\infty}}.
$$

Essa estimativa mostra que a espessura da camada-limite cresce aproximadamente com $\sqrt{x}$.

Assim, podemos definir uma coordenada vertical reescalada por

$$
\eta
=
\frac{y}{\delta(x)}
=
y\sqrt{\frac{U_\infty}{\nu x}}.
$$

A variável $\eta$ permite comparar perfis localizados em diferentes posições $x$ usando uma única coordenada.

### **Função corrente e transformação de similaridade**

Para satisfazer automaticamente a equação de conservação de massa, introduzimos uma função corrente $\psi(x,y)$:

$$
u=\frac{\partial\psi}{\partial y},
\qquad
v=-\frac{\partial\psi}{\partial x}.
$$

A solução de Blasius propõe a seguinte forma para essa função:

$$
\psi(x,y)
=
\sqrt{\nu U_\infty x}\,f(\eta),
$$

em que $f$ é uma função desconhecida apenas da variável de similaridade $\eta$.

A partir dessa definição, a velocidade na direção principal do escoamento torna-se

$$
u
=
U_\infty f'(\eta),
$$

ou, de forma adimensional,

$$
\frac{u}{U_\infty}
=
f'(\eta).
$$

A componente vertical da velocidade é dada por

$$
v
=
\frac{1}{2}
\sqrt{\frac{\nu U_\infty}{x}}
\left[
\eta f'(\eta)-f(\eta)
\right].
$$

Portanto, a dependência original da velocidade em $x$ e $y$ passa a ser representada por uma única função $f(\eta)$.

### **Equação de Blasius**

Substituindo a transformação de similaridade na equação de quantidade de movimento, obtemos a seguinte equação diferencial ordinária:

$$
f'''(\eta)
+
\frac{1}{2}f(\eta)f''(\eta)
=
0.
$$

As condições de contorno são obtidas diretamente das condições físicas do problema:

$$
f(0)=0,
$$

associada à condição de não penetração na parede;

$$
f'(0)=0,
$$

associada à condição de não deslizamento; e

$$
f'(\eta)\longrightarrow 1
\qquad
\text{quando } \eta\longrightarrow\infty,
$$

associada à recuperação da velocidade do escoamento externo.

Assim, um problema originalmente descrito por uma equação diferencial parcial em $x$ e $y$ foi reduzido a uma equação diferencial ordinária em $\eta$.

No código, resolveremos essa equação pelo método de shooting. O valor desconhecido será $f''(0)$.

A integração será iniciada com

$$
f(0)=0,
\qquad
f'(0)=0,
\qquad
f''(0)=\text{chute}.
$$

O chute será ajustado até que a condição

$$
f'(10)\approx 1
$$

seja satisfeita. O valor $\eta=10$ representa numericamente o limite $\eta\to\infty$.

## **Como a equação de Blasius é resolvida numericamente?**

A equação de Blasius é uma equação diferencial ordinária de terceira ordem. Para utilizar um integrador numérico, transformamos essa equação em um sistema de três equações de primeira ordem.

Definimos:

$$y_1=f, \qquad y_2=f', \qquad y_3=f''.$$

Com isso, a equação passa a ser escrita como:

$$\frac{dy_1}{d\eta}=y_2,$$

$$\frac{dy_2}{d\eta}=y_3,$$

$$\frac{dy_3}{d\eta}= -\frac{1}{2}y_1y_3.$$

Na parede, conhecemos:

$$f(0)=0, \qquad f'(0)=0.$$

Entretanto, o valor de $f''(0)$ não é conhecido. Esse valor representa o gradiente de velocidade na parede e precisa ser determinado.

O **método de shooting** resolve esse problema da seguinte forma:

1. escolhemos um valor inicial para $f''(0)$;
2. integramos o sistema até um valor elevado de $\eta$;
3. verificamos o valor obtido para $f'(\eta)$;
4. ajustamos o chute até que $f'(\eta)$ se aproxime de 1.

O valor $\eta=10$ é utilizado como uma aproximação numérica para $\eta\to\infty$.

### **Observação sobre a integração numérica**

A função `solve_ivp` realiza a integração do sistema de equações diferenciais utilizando um método numérico baseado em Runge–Kutta com controle adaptativo do passo.

Neste notebook, não implementaremos manualmente todos os passos do método. O objetivo principal é compreender como o sistema é construído e como o método de shooting determina o valor desconhecido de $f''(0)$.

A implementação detalhada de Euler, Runge–Kutta e dos métodos de busca de raízes é apresentada separadamente.

````python
import numpy as np
from scipy.integrate import solve_ivp
from scipy.optimize import root_scalar
from scipy.interpolate import CubicSpline
import matplotlib.pyplot as plt
````

````python
# 1. Definição do sistema de EDOs (Redução de Ordem)
def blasius_ode(eta, y):
    """
    Sistema de primeira ordem associado à equação de Blasius.
    Equação original:
        f''' + 0.5*f*f'' = 0

    Variáveis:
        y[0] = f, y[1] = f', y[2] = f''

    Retorno:
        [f', f'', f''']
    """
    f, df, d2f = y

    return [df, d2f, -0.5 * f * d2f,]
````

````python
# 2. Função Objetivo (O "Alvo")
def objective_function(guess):
    """
    Roda a integração para um dado 'chute' de f''(0).
    Retorna o erro em relação à condição de contorno no infinito.
    """
    # Condições iniciais na parede (eta = 0): f=0, f'=0, f''=guess
    y0 = [0.0, 0.0, guess]

    # A condição física no escoamento externo é:
    # f'(infinito) = 1
    #
    # Como não podemos integrar até o infinito, usamos eta = 10 como aproximação.
    solver_options = {"method": "RK45", "rtol": 1e-9, "atol": 1e-11,}
    sol = solve_ivp(blasius_ode, [0, 10], y0, **solver_options)
    f_prime_inf = sol.y[1][-1]
    error = f_prime_inf - 1.0

    return error
````

**testando valores distintos para f''(0)**

````python
# Testando alguns chutes para f''(0)
chutes = np.array([0.10, 0.20, 0.30, 0.332057, 0.35, 0.40, 0.50, 1.0])

print(" Chute para f''(0)     f'(10)        Erro")
print("-" * 46)

for chute in chutes:
    erro = objective_function(chute)
    valor_final = erro + 1.0

    print(
        f"      {chute:8.6f}       "
        f"{valor_final:8.6f}     "
        f"{erro: .6e}"
    )
````

#### **Usando um método de busca de raízes**

O objetivo não é minimizar uma função qualquer, mas encontrar o valor de $f''(0)$ que faz o erro final desaparecer.

O método `brentq` procura a raiz da função

$$
R(\alpha)=f'(10;\alpha)-1,
$$

em que $\alpha$ representa o chute utilizado para $f''(0)$.

````python
# 3. Disparo automatizado (Encontrando a raiz)
# Vamos dizer que a segunda derivada na parede está entre 0.1 e 1.0
print("Buscando o valor de f''(0) pelo método de shooting...")
root_result = root_scalar(objective_function, bracket=[0.1, 1.0], method='brentq')

# O chute perfeito que zera o erro:
best_guess = root_result.root
print(f"Valor convergido para f''(0): {best_guess:.5f}")
````

### **Solução Final e Geração de Gráficos**

````python
# 4. Solução Final e Geração de Gráficos
# Roda a integração uma última vez com o chute correto para gerar os dados contínuos
eta_span = [0, 10]
eta_eval = np.linspace(0, 10, 500) # Vetor denso para um gráfico suave
solver_options = {"method": "RK45", "rtol": 1e-9, "atol": 1e-11,}
sol_final = solve_ivp(blasius_ode, eta_span, [0.0, 0.0, best_guess],
                      t_eval=eta_eval, **solver_options,)

eta = sol_final.t
f = sol_final.y[0]
df = sol_final.y[1]   # Esta é a variável atrelada à velocidade u
d2f = sol_final.y[2]  # Esta é a variável atrelada ao cisalhamento

fig, axes = plt.subplots(1, 3, figsize=(10, 3))

axes[0].plot(f, eta, color="tab:red", linewidth=2, label=r"$f(\eta)$",)
#axes[0].set_title("Perfil de velocidade")
axes[0].set_xlabel(r"$f$")
axes[0].set_ylabel(r"$\eta$")
axes[0].grid(True, linestyle=":")
axes[0].legend()

axes[1].plot(df, eta, color="tab:blue", linewidth=2, label=r"$u/U_\infty=f'(\eta)$",)
axes[1].set_title("Perfil de velocidade")
axes[1].set_xlabel(r"$u/U_\infty$")
axes[1].set_ylabel(r"$\eta$")
axes[1].set_xlim(0, 1.05)
axes[1].grid(True, linestyle=":")
axes[1].legend()

axes[2].plot(d2f, eta, color="tab:green",linewidth=2,label=r"$f''(\eta)$",)
axes[2].set_title("Gradiente de velocidade")
axes[2].set_xlabel(r"$f''(\eta)$")
axes[2].set_ylabel(r"$\eta$")
axes[2].grid(True, linestyle=":")
axes[2].legend()

plt.tight_layout()
plt.show()
````

````python
# Parâmetros físicos ilustrativos
U_inf = 10.0       # velocidade externa [m/s]
nu = 1.5e-5        # viscosidade cinemática [m²/s]

x_posicoes = [0.1, 0.5, 1.0]

fig, ax = plt.subplots(figsize=(8, 3))

for x in x_posicoes:
    y = eta * np.sqrt(nu * x / U_inf)

    ax.plot(
        y * 1000,
        df,
        linewidth=2,
        label=fr"$x={x:.1f}\ \mathrm{{m}}$",
    )

ax.set_xlabel(r"Distância da parede $y$ [mm]")
ax.set_ylabel(r"Velocidade normalizada $u/U_\infty$")
ax.set_title("Desenvolvimento da camada-limite ao longo da placa")
ax.grid(True, linestyle=":")
ax.legend()
plt.show()
````

### **Crescimento da camada-limite ao longo da placa**

A camada-limite hidrodinâmica começa praticamente com espessura nula na borda de ataque e aumenta à medida que o fluido percorre a placa.

Usando o critério de 99%, definimos sua espessura por:

$$\frac{u}{U_\infty}=0.99.$$

A espessura correspondente é dada por:

$$\delta_{99}(x) = \eta_{99}\sqrt{\frac{\nu x}{U_\infty}}.$$

Essa relação mostra que a camada-limite cresce aproximadamente com $\sqrt{x}$.

Uma placa mais longa não altera a espessura da camada em uma mesma posição $x$. Ela apenas permite que o desenvolvimento da camada continue por uma distância maior.

````python
# Comprimentos totais de placa analisados
comprimentos_placa = [0.25, 0.50, 1.00]

# Valor de eta associado a u/U_inf = 0.99
eta_99 = np.interp(0.99, df, eta)

# Espessura máxima para definir o limite vertical comum
delta_max = eta_99 * np.sqrt(nu * max(comprimentos_placa) / U_inf)

fig, axes = plt.subplots(1,3,figsize=(15, 5),sharey=True,)

for ax, L_plate in zip(axes, comprimentos_placa):
    x_plot = np.linspace(0.005, L_plate, 300)
    delta_99 = eta_99 * np.sqrt(nu * x_plot / U_inf)

    # Camada-limite hidrodinâmica
    ax.fill_between(x_plot, 0, delta_99 * 1000, color="tab:blue", alpha=0.25,)
    ax.plot(x_plot, delta_99 * 1000, color="tab:blue", linewidth=2,
        label=r"$\delta_{99}(x)$",)

    # Representação da placa
    ax.fill_between(x_plot, -0.25, 0, color="saddlebrown", alpha=0.85,)
    ax.axhline(0, color="black", linewidth=1,)
    ax.set_title(fr"Placa com $L={L_plate:.2f}\,\mathrm{{m}}$")
    ax.set_xlabel(r"$x$ [m]")
    ax.grid(True, linestyle=":")
    ax.legend()

    delta_final = delta_99[-1]

    print(f"L = {L_plate:.2f} m  -->  "
        f"delta_99(L) = {delta_final * 1000:.3f} mm")

axes[0].set_ylabel(r"Espessura da camada-limite [mm]")

for ax in axes:
    ax.set_ylim(-0.4,delta_max * 1000 * 1.15,)

fig.suptitle(
    "Influência do comprimento da placa na camada-limite hidrodinâmica",
    fontsize=14,
)

plt.tight_layout()
plt.show()
````

Nos três casos, a tendência de crescimento é a mesma. A diferença é que uma placa mais longa permite que o fluido percorra uma distância maior e, consequentemente, que a camada-limite se desenvolva por mais tempo.

Assim, a espessura ao final da placa aumenta com o comprimento $L$, mas a relação local $\delta_{99}(x)$ permanece determinada pela posição $x$, pela viscosidade cinemática e pela velocidade externa.

# **Da camada hidrodinâmica à camada térmica**

Depois de determinar o perfil de velocidade, podemos analisar a distribuição de temperatura próxima à placa.

Consideraremos uma parede mantida a temperatura constante $T_w$ e um escoamento externo à temperatura $T_\infty$.

A variável adimensional de temperatura será definida como

$$
\theta(\eta) = \frac{T-T_\infty}{T_w-T_\infty}.
$$

Assim:

- $\theta=1$ na parede;
- $\theta=0$ longe da placa.

A camada térmica é influenciada pelo perfil de velocidade obtido na solução de Blasius.



## **Equação da camada-limite térmica**

Com as mesmas hipóteses utilizadas para a camada hidrodinâmica, a equação da energia pode ser reduzida a uma equação diferencial ordinária para $\theta(\eta)$:

$$\theta'' + \frac{Pr}{2}f(\eta)\theta' = 0,$$

em que $Pr$ é o número de Prandtl:

$$Pr=\frac{\nu}{\alpha}.$$

As condições de contorno são:

$$\theta(0)=1, \qquad \theta(\infty)=0.$$

O valor de $Pr$ compara a difusão de quantidade de movimento com a difusão térmica. Por isso, ele influencia diretamente a espessura da camada térmica.

### **Procedimento de Solução Numérica**

Ideia geral:

```
já conhecemos f(η)
        ↓
usamos f(η) para resolver θ(η)
        ↓
descobrimos qual inclinação θ'(0) faz θ(η) chegar a zero
```

---

Anteriormente, resolvemos a equação de Blasius e obtivemos vários valores:

$$\eta = [0.0, 0.02, 0.04, ...]$$

$$f(\eta)   = [0.0, 0.00, 0.01, ...]$$

Ou seja, temos valores de $f$ apenas em alguns pontos. O problema é que o solve_ivp pode perguntar:
> "Qual é o valor de $f$ em $\eta=2.347$?”

Para responder, criamos uma função contínua aproximada:

````python
# Interpolação do campo hidrodinâmico f(eta)
# Esse campo foi obtido na solução de Blasius.
f_interp = CubicSpline(eta, f)
````

````python
plt.subplots(figsize=(8, 3))
plt.plot(eta, f,'-', label = 'Valores Iniciais')
plt.plot(eta, f_interp(eta),'--', label = 'Interpolacao')
plt.xlabel(r'$\eta$')
plt.ylabel('f')
plt.grid()
plt.legend()
plt.show()
````

Agora desenvolvemos equação térmica:

$$
\theta'' + \frac{Pr}{2}f(\eta)\theta'=0.
$$

Os integradores de EDO utilizados pelo `solve_ivp` recebem o problema na forma de um sistema de equações de primeira ordem. Por isso, uma equação de segunda ordem precisa ser reescrita introduzindo $\theta$ e $\theta'$ como variáveis independentes. Então definimos:
$$ y_0=\theta, \qquad y_1=\theta'. $$

Assim:
$$\frac{dy_0}{d\eta}=y_1,$$

e, reorganizando a equação térmica:

$$
\frac{dy_1}{d\eta} = \theta'' = -\frac{Pr}{2}f(\eta)y_1.
$$

````python
def thermal_ode(eta_local, y, Pr):
    """
    Sistema de primeira ordem para a camada térmica.
    y[0] = theta
    y[1] = theta'
    """
    theta, dtheta = y
    return [dtheta,
            -0.5 * Pr * f_interp(eta_local) * dtheta,]
````

**Qual é o problema?**

Na parede sabemos:
$$\theta(0)=1.$$

Mas **não sabemos a inclinação inicial**:

$$\theta'(0)=?$$

Essa inclinação é importante. Se ela for **pouco negativa**, a temperatura demora muito para cair. Se for **muito negativa**, a temperatura cai rápido demais.

---

**Por que $\theta'(0)$ deve ser negativo?**

A coordenada $\eta$ aumenta à medida que nos afastamos da parede. Como a temperatura adimensional vale $\theta=1$ na parede e tende a zero longe dela, o perfil térmico deve diminuir ao longo de $\eta$.

Por isso,

$$ \theta'(0)< 0.$$

O sinal resulta da condição física de que a temperatura deve variar de $T_w$ até $T_\infty$.

---

A função abaixo testa um chute para essa inclinação:

````python
def thermal_residual(slope, Pr):
    """
    Retorna o valor de theta no final do domínio
    para um determinado chute de theta'(0).
    """
    sol = solve_ivp(thermal_ode, [0, 10], [1.0, slope], args=(Pr,),)
    return sol.y[0, -1]
````

Esta parte: `[1.0, slope]` representa as condições iniciais `theta(0) = 1, theta'(0) = slope`

Por exemplo, se slope = -0.3 o integrador começa com:
$$\theta(0)=1, \qquad \theta'(0)=-0.3. $$

Depois ele integra até $\eta=10$. No final, observamos: `sol.y[0, -1]`

Esse valor é simplesmente $\theta(10)$. Como desejamos que a temperatura volte à temperatura externa longe da parede, queremos:

$$\theta(10)\approx 0.$$

Por isso, chamamos esse valor de **residual ou erro**.



#### **Método Shooting**

````python
def solve_thermal_profile(Pr):
    """
    Resolve o problema térmico pelo método de shooting.
    """

    # theta'(0) é desconhecido e deve ser negativo.
    root = root_scalar(lambda slope: thermal_residual(slope, Pr),
                       bracket=[-2.0, -0.01], method="brentq",)

    slope_wall = root.root

    sol = solve_ivp(thermal_ode, [0, 10],
        [1.0, slope_wall], t_eval=eta, args=(Pr,),)

    return slope_wall, sol.y[0]
````

**comparação entre diferentes números de Prandtl**

````python
Pr_values = [0.6, 1.0, 7.0]

thermal_profiles = {}
thermal_slopes = {}

for Pr in Pr_values:
    slope_wall, theta = solve_thermal_profile(Pr)

    thermal_slopes[Pr] = slope_wall
    thermal_profiles[Pr] = theta

    print(
        f"Pr = {Pr:.1f}  |  "
        f"theta'(0) = {slope_wall:.6f}"
    )
````

````python
fig, ax = plt.subplots(figsize=(8, 3))

for Pr in Pr_values:
    ax.plot(
        thermal_profiles[Pr],
        eta,
        linewidth=2,
        label=fr"$Pr={Pr}$",
    )

ax.set_xlabel(r"$\theta(\eta)$")
ax.set_ylabel(r"$\eta$")
ax.set_ylim([0,10])
ax.set_title("Perfis adimensionais de temperatura")
ax.grid(True, linestyle=":")
ax.legend()

plt.show()
````

#### **Exemplo com temperatura dimensional**

````python
# Exemplo de temperatura dimensional
Pr_ref = 0.6
T_wall = 400.0   # [K]
T_inf = 300.0    # [K]
x_ref = 0.5      # posição ao longo da placa [m]

theta_ref = thermal_profiles[Pr_ref]

y_thermal = eta * np.sqrt(nu * x_ref / U_inf)
T = T_inf + theta_ref * (T_wall - T_inf)

plt.figure(figsize=(8, 3))
plt.plot(y_thermal * 1000, T, linewidth=2)

plt.xlabel("Distância da parede [mm]")
plt.ylabel("Temperatura [K]")
plt.title(
    fr"Perfil de temperatura em $x={x_ref}\,\mathrm{{m}}$ "
    fr"($Pr={Pr_ref}$)"
)
plt.grid(True, linestyle=":")
plt.show()
````

## **Comparação entre as camadas hidrodinâmica e térmica**

A camada hidrodinâmica é representada pelo perfil

$$
\frac{u}{U_\infty}=f'(\eta),
$$

enquanto a camada térmica é representada por

$$
\theta(\eta).
$$

Os dois perfis são comparados para diferentes valores do número de Prandtl.

A curva hidrodinâmica é a mesma em todos os casos, pois o escoamento não foi alterado. Já a curva térmica muda conforme o valor de $Pr$.

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 5),
    sharex=True,
    sharey=True,
)

for ax, Pr in zip(axes, Pr_values):
    theta = thermal_profiles[Pr]

    # Perfil hidrodinâmico
    ax.plot(df, eta,
        color="tab:blue",
        linewidth=2,
        label=r"$u/U_\infty=f'(\eta)$",
    )

    # Perfil térmico
    ax.plot(
        theta,
        eta,
        color="tab:red",
        linewidth=2,
        label=r"$\theta(\eta)$",
    )

    ax.set_title(fr"$Pr={Pr}$")
    ax.set_xlabel("Valor adimensional")
    ax.grid(True, linestyle=":")

axes[0].set_ylabel(r"$\eta$")
axes[0].set_xlim(0, 1.05)
axes[0].set_ylim(0, 10)

axes[0].legend()

fig.suptitle(
    "Comparação entre as camadas hidrodinâmica e térmica",
    fontsize=14,
)

plt.tight_layout()
plt.show()
````

### **Espessura das camadas ao longo da placa**

Os perfis anteriores foram apresentados em função da coordenada de similaridade $\eta$. Para visualizar a camada-limite como uma região física, precisamos calcular sua espessura ao longo da placa. Adotaremos as seguintes definições convencionais:

- $\delta_h$: posição em que $u/U_\infty=0.99$;
- $\delta_T$: posição em que $\theta=0.01$.

A espessura física é obtida a partir de

$$
y=\eta\sqrt{\frac{\nu x}{U_\infty}}.
$$

Como consequência, tanto a camada hidrodinâmica quanto a camada térmica crescem aproximadamente com $\sqrt{x}$.

````python
# Domínio ao longo da placa
x_plot = np.linspace(0.02, 1.0, 300)

# Valor de eta associado à espessura hidrodinâmica
eta_h_99 = np.interp(0.99, df, eta)

# Espessura hidrodinâmica ao longo da placa
delta_h = eta_h_99 * np.sqrt(nu * x_plot / U_inf)

fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharex=True,sharey=True,)

for ax, Pr in zip(axes, Pr_values):
    theta = thermal_profiles[Pr]

    # Valor de eta associado à espessura térmica
    # Como theta diminui, usamos os vetores invertidos.
    eta_T_99 = np.interp(
        0.01,
        theta[::-1],
        eta[::-1],
    )

    # Espessura térmica ao longo da placa
    delta_T = eta_T_99 * np.sqrt(nu * x_plot / U_inf)

    # Camada hidrodinâmica
    ax.fill_between(
        x_plot,
        0,
        delta_h * 1000,
        color="tab:blue",
        alpha=0.25,
        label=r"camada hidrodinâmica $\delta_h$",
    )

    # Camada térmica
    ax.fill_between(
        x_plot,
        0,
        delta_T * 1000,
        color="tab:red",
        alpha=0.25,
        label=r"camada térmica $\delta_T$",
    )

    # Fronteiras das camadas
    ax.plot(
        x_plot,
        delta_h * 1000,
        color="tab:blue",
        linewidth=2,
    )

    ax.plot(
        x_plot,
        delta_T * 1000,
        color="tab:red",
        linewidth=2,
    )

    # Representação da placa
    ax.fill_between(
        x_plot,
        -0.25,
        0,
        color="saddlebrown",
        alpha=0.85,
    )

    ax.axhline(0, color="black", linewidth=1)

    ax.set_title(fr"$Pr={Pr}$")
    ax.set_xlabel(r"$x$ [m]")
    ax.grid(True, linestyle=":")

axes[0].set_ylabel(r"Distância da parede [mm]")
axes[0].legend(fontsize=8, loc="upper left")

fig.suptitle(
    "Crescimento das camadas hidrodinâmica e térmica",
    fontsize=14,
)

plt.tight_layout()
plt.show()
````

A camada hidrodinâmica é a mesma nos três casos, pois o campo de velocidade não depende do número de Prandtl.

A camada térmica, por outro lado, muda de espessura:

- para $Pr < 1$, a camada térmica tende a ser mais espessa;
- para $Pr = 1$, as duas camadas possuem espessuras semelhantes;
- para $Pr > 1$, a camada térmica tende a ser mais fina.

Essa representação mostra que as camadas não são fronteiras abruptas. Elas são regiões próximas à parede nas quais a velocidade ou a temperatura variam significativamente até se aproximarem dos valores externos.
